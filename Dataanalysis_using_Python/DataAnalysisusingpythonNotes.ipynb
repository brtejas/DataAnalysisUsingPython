### Different Types of seaborn plots() available in EDA

#### 1) Distribution plots
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

##### 1. histplot():

sns.histplot(tips['total_bill'])

plt.show()

Use when you want to see frequency distribution of a variable

##### 2. kdeplot():

sns.kdeplot(tips['total_bill'])

plt.show()

Use when you want smooth distribution curve

##### 3. displot():
   
sns.displot(tips['total_bill'])

plt.show()

Use when you want combined histogram + KDE

##### 4. ecdfplot():
   
sns.ecdfplot(tips['total_bill'])

plt.show()

Use when you want cumulative distribution

#### 2) Categorical plots

##### 1. countplot():

sns.countplot(x='day', data=tips)

plt.show()

Use when you want count of categories

##### 2. barplot():

sns.barplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want average values across categories

##### 3. boxplot():

sns.boxplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want spread + outliers of data

##### 4. voilinplot():

sns.violinplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want distribution + density of categories

##### 5. stripplot():

sns.stripplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want raw data points visualization

##### 6. swarmplot():

sns.swarmplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want non-overlapping data points

##### 7. pointplot():
   
sns.pointplot(x='day', y='total_bill', data=tips)

plt.show()

Use when you want trend of mean values

#### 3) Relational plots

##### 1. scatterplot():
   
sns.scatterplot(x='total_bill', y='tip', data=tips)

plt.show()

Use when you want relationship between two numeric variables

##### 2. lineplot():

sns.lineplot(x='total_bill', y='tip', data=tips)

plt.show()

Use when you want trend over continuous data

##### 3. relplot():

sns.relplot(x='total_bill', y='tip', data=tips)

plt.show()

Use when you want flexible relational plotting with facets

#### 4) Matrix plots

##### 1. heatmap():

corr = tips.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.show()

Use when you want correlation between variables

##### 2. clustermap():
   
sns.clustermap(corr)

plt.show()

Use when you want clustered correlation patterns

#### 5) Regression plots
##### 1. regplot():
   
sns.regplot(x='total_bill', y='tip', data=tips)

plt.show()

Use when you want regression trend line

##### 2. lmplot():

sns.lmplot(x='total_bill', y='tip', data=tips)

plt.show()

Use when you want regression with grouping support

### Different Types of connection code (conn) to connect various datasources using python

#### 1) MySQL

import mysql.connector

conn = mysql.connector.connect(host="localhost", user="root", password="1234", database="mydb")

Use when connecting to MySQL databases

#### 2) PostgreSQL

import psycopg2

conn = psycopg2.connect(host="localhost", user="postgres", password="1234", dbname="mydb")

Use when connecting to PostgreSQL databases

#### 3) SQLite

import sqlite3

conn = sqlite3.connect("mydatabase.db")

Use when working with lightweight local database files

#### 4) SQL Server

import pyodbc

conn = pyodbc.connect("DRIVER={SQL Server};SERVER=localhost;DATABASE=mydb;UID=user;PWD=1234")

Use when connecting to Microsoft SQL Server

#### 5) Oracle

import cx_Oracle

conn = cx_Oracle.connect("user/1234@localhost:1521/orcl")

Use when connecting to Oracle databases

#### 6) MongoDB (NoSQL)

from pymongo import MongoClient

conn = MongoClient("mongodb://localhost:27017/")

Use when working with NoSQL MongoDB databases

#### 7) Snowflake

import snowflake.connector

conn = snowflake.connector.connect(user='user', password='1234', account='account')

#### 8) AWS Redshift

import psycopg2

Use when connecting to Amazon Redshift warehouse

#### 9) Google BigQuery

Use when reading Excel files

#### 11) CSV file

import requests

Use when fetching data from APIs

### Different Methods of reading the data from the data sources (pd.read_csv()) 

#### 1. TEXT / FILE FORMATS

import pandas as pd

##### CSV

df = pd.read_csv("data.csv")

Use when data is stored in comma-separated file

##### Excel

df = pd.read_excel("data.xlsx")

Use when working with Excel sheets

##### JSON

df = pd.read_json("data.json")

Use when data is in JSON format

##### Text file

df = pd.read_table("data.txt")

Use when data is tab-separated

##### Fixed-width file

df = pd.read_fwf("data.txt")

Use when columns are aligned by spacing

#### 2. DATABASE SOURCES

##### SQL

df = pd.read_sql("SELECT * FROM table", conn)

Use when reading from SQL databases

##### SQL Table directly

df = pd.read_sql_table("table_name", conn)

Use when reading full table from database

##### SQL Query

df = pd.read_sql_query("SELECT * FROM table", conn)

Use when executing custom SQL queries

#### 3. ONLINE / WEB DATA

##### From URL (CSV)

df = pd.read_csv("https://example.com/data.csv")

Use when reading CSV directly from internet

##### HTML tables

df = pd.read_html("https://example.com")[0]

Use when extracting tables from web pages

#### 4. SPECIAL FORMATS

##### Parquet

df = pd.read_parquet("data.parquet")

Use when working with optimized big data format

##### Feather

df = pd.read_feather("data.feather")

Use when fast read/write binary format needed

##### Pickle

df = pd.read_pickle("data.pkl")

Use when loading previously saved pandas objects

##### HDF5
df = pd.read_hdf("data.h5")

Use when handling large hierarchical datasets

##### CLIPBOARD (quick trick)

df = pd.read_clipboard()

Use when copying table data directly from screen

##### One-line memory

Every data format has its own pd.read_* function

### Different Methods of writing the data into their respective data sources back.

#### TEXT / FILE FORMATS

import pandas as pd

##### CSV

df.to_csv("output.csv", index=False)

Use when saving data in simple comma-separated format

##### Excel

df.to_excel("output.xlsx", index=False)

Use when exporting data to Excel file

##### Multiple sheets (Excel)

with pd.ExcelWriter("output.xlsx") as writer:

df.to_excel(writer, sheet_name="Sheet1", index=False)
    
Use when writing multiple sheets in one Excel file

##### JSON

df.to_json("output.json", orient="records")

Use when exporting data in JSON format

##### Text / TSV

df.to_csv("output.txt", sep="\t", index=False)

Use when saving tab-separated file

#### DATABASE SOURCES

##### SQL table

df.to_sql("table_name", conn, if_exists="replace", index=False)

Use when writing data into SQL database table

#### SPECIAL FORMATS

##### Parquet

df.to_parquet("output.parquet")

Use when saving large data efficiently

##### Feather

df.to_feather("output.feather")

Use when fast read/write is required

##### Pickle

df.to_pickle("output.pkl")

Use when saving pandas object 

for later reuse

##### HDF5
df.to_hdf("output.h5", key="df", mode="w")

Use when storing large hierarchical datasets

##### WEB / API (manual way)

import requests

requests.post("https://api.example.com/upload", json=df.to_dict())

Use when sending data to API

##### CLIPBOARD

df.to_clipboard(index=False)

Use when copying data directly to clipboard

##### One-line memory

Every pd.read_* has a matching df.to_* for writing data



### Different Methods or types of using pandas to create new columns

#### 1) Direct assignment (MOST COMMON)

df['New'] = df['Amount'] * 2

Use when applying simple arithmetic

####  2) Using .loc[]

df.loc[:, 'New'] = df['Amount'] * 2

Use when avoiding SettingWithCopyWarning

#### 3) Using .assign()

df = df.assign(New=df['Amount'] * 2)

Use when chaining operations

#### 4) Using .apply() (row-wise)

df['New'] = df.apply(lambda row: row['A'] + row['B'], axis=1)

Use when logic involves multiple columns

#### 5) Using np.where() (conditions)

import numpy as np

df['New'] = np.where(df['Amount'] > 500, 'High', 'Low')

Use for IF condition (fast)

#### 6) Using np.select() (multiple conditions)

df['New'] = np.select(

    [df['Amount'] > 500, df['Amount'] > 200],
    
    ['High', 'Medium'],
    
    default='Low'
    
)

Use for multiple IF conditions

#### 7) Using .map() / .replace()

df['New'] = df['Gender'].map({'M': 'Male', 'F': 'Female'})

Use for value mapping

#### 8) Using groupby().transform() (advanced)

df['Avg'] = df.groupby('State')['Amount'].transform('mean')

Use for group-based calculations (like DAX)

#### 9) Using string operations

df['Full'] = df['First'] + " " + df['Last']

Use for text-based columns

#### 10) Using datetime operations

df['Year'] = pd.to_datetime(df['Date']).dt.year

Use for date extraction
