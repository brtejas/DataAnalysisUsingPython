```python
#importing the libraries
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#loading the data
df = pd.read_csv(r"C:\Users\User\Downloads\DATA-ANALYSIS\Dataanalysis_using_Python\Python_VideoGamesProject-main\VideoGamesSales.csv")
df
```

```python
#Knowing the size of the dataset
df.shape
```

```python
#Knowing the duplicated values and removing them in the dataset
#df[df.duplicated()]
df = df.drop_duplicates()
df.shape
```

```python
#Knowing the information about the columns
df.info()
```

```python
pd.isnull(df)
```

```python
pd.isnull(df).sum()
```

```python

df.loc[:, 'Region'] = df['Region'].fillna('North')
pd.isnull(df).sum()
```

```python
df.loc[:,'Publisher'] = df['Publisher'].fillna('Nintendo')
pd.isnull(df).sum()
```

```python
#top 100 values if you want df.head()
df.head(100)
```

```python
#if you want last 100 values then df.tail()
df.tail(100)
```

```python
# Replacing '$' with '' using regex (regular expression)
df.loc[:,'NA_Sales'] = df['NA_Sales'].replace('[$]','', regex = True)
df
```

```python
# Converting NA_Sales column dtype from alphanumeric to numeric using pandas
df.loc[:, 'NA_Sales'] = pd.to_numeric(df['NA_Sales'], errors = 'coerce')
df
```

```python
#Finding the average Sales using mean() function
Average_Sales = df['NA_Sales'].mean()
Average_Sales
```

```python
# Filling the average sales value from  0 to 0.65
df['NA_Sales'] = df['NA_Sales'].replace(0,Average_Sales)
df
```

```python
# Replacing the value in country from USA to United States
df.loc[:, 'Country'] = df['Country'].replace({'USA' : 'United States'})
df
```

```python
# Converting the value to have capital letter
df.loc[:,'Country'] = df['Country'].str.title()
df
```

```python
# if you want the number of entries with a particular value
#df = df[df['Country'] == 'United States']
#df
```

```python
# Renaming the columns from old to new values
df = df.rename(columns = {'NA_Sales':'National Sales', 'Global_Sales' : 'Global Sales', 'NA_Profit':'National Profit', 
                          'Global_Profit' : 'Global Profit'})
df
```

```python
# Create the bar chart 
National_Sales = df.groupby(['Region','Country']) ['National Sales'].sum().reset_index().sort_values(by = 'National Sales', ascending = False)
#sns.set(rc={'figure.figsize':(20,5)})
plt.figure(figsize = (20,5))
sns.barplot(data = National_Sales, x = 'Region', y = 'National Sales', hue = 'Country')
plt.title("National Sales by Region", fontweight='bold')
plt.xlabel("Region", fontweight='bold')
plt.ylabel("National Sales",fontweight='bold')
plt.show()
```

```python
#Create a pie chart
Sales = df.groupby(['Country']) [['National Sales','Global Sales']].sum().reset_index()
Country = Sales['Country']
National_Sales = Sales['National Sales']
Global_Sales = Sales['Global Sales']

# (1,2)-> (rows, number of charts)
fig, axs = plt.subplots(1,2, figsize = (16,8))

axs[0].pie(National_Sales, labels = Country, autopct = '%1.1f%%', startangle = 90)


axs[1].pie(National_Sales, labels = Country, autopct = '%1.1f%%', startangle = 120)

plt.show()
```

```python
Sales = df.groupby(['Country']) [['National Sales', 'Global Sales']].sum().reset_index()
Country = Sales['Country']
National_Sales = Sales['National Sales']
Global_Sales = Sales['Global Sales']

fig, axs = plt.subplots(1,2,figsize = (10,7))

axs[0].pie(National_Sales, labels = Country, autopct = "%1.1f%%", startangle = 90)
axs[0].set_title("National Sales by Country")

axs[1].pie(Global_Sales, labels = Country, autopct = "%1.1f%%", startangle = 140)
axs[1].set_title("Global Sales by Country")

plt.show()
```
